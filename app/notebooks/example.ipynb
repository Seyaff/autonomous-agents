```python
from langchain_core.prompts import PromptTemplate

# Define template with placeholders
template = PromptTemplate.from_template("Translate the following text to {language}: {text}")

# Format it manually
formatted_prompt = template.format(language="French", text="Hello world!")
print(formatted_prompt)
# Output: Translate the following text to French: Hello world!
```

```python
import os
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq


load_dotenv()

# 1. Initialize the ChatGroq model
# (Popular Groq models include llama-3.3-70b-versatile, llama3-8b-8192, mixtral-8x7b-32768)
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.7,
    # api_key="gsk_..." # Optional if set in os.environ["GROQ_API_KEY"]
)

# 2. Define a ChatPromptTemplate with role tuples and variable placeholders
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role} who explains concepts simply and concisely."),
    ("human", "Explain {topic} in 2-3 sentences.")
])

# 3. Output parser to convert the AIMessage response directly into plain text
output_parser = StrOutputParser()

# 4. Chain components together using the pipe (|) operator (LCEL)
chain = prompt | llm | output_parser

# 5. Invoke the chain passing values for the prompt variables
response = chain.invoke({
    "role": "data science mentor",
    "topic": "quantum computing"
})

print(response)
```
